```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_squared_error as mse, mean_absolute_error as mae, r2_score
```

```python
df = pd.read_csv('./Data.csv')
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df[df['Output']==None]
```

```python
df[df['Output'].isnull()]
```

```python
df.dropna(inplace = True)
```

```python
df.iloc[968]
```

```python
X = df.drop(columns = ['Output']).to_numpy()
y = df['Output'].to_numpy()
```

```python
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.15, random_state = 42)
X_train, y_train, X_test, y_test
```

```python
model = RandomForestRegressor(
    n_estimators = 100,
    max_depth = 4,
    random_state = 42
)
model.fit(X_train,y_train)
```

```python
# import pickle

# with open('./App/Models/RFReg.pkl','wb') as file:
#     pickle.dump(model, file)
#     print('pickle dump successful')
```

```python
print(mse(y_train, model.predict(X_train)))
print(mae(y_train, model.predict(X_train)))
print(r2_score(y_train, model.predict(X_train)))
```

```python
print(mse(y_test, model.predict(X_test)))
print(mae(y_test, model.predict(X_test)))
print(r2_score(y_test, model.predict(X_test)))
```
