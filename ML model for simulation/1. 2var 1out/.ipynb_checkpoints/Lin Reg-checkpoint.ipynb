```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.read_csv('cleaned data.csv')
df.head()
```

```python
df.describe()
```

```python
plt.scatter(df['variable 2'].iloc[:25],df['Output'].iloc[:25])
```

```python
plt.scatter(df['variable 2'].iloc[:25],np.log(df['Output'].iloc[:25]))
```

Adding Log transform to linearize data

```python
X = df.drop(columns = ['Output'])
Y = np.log(df['Output'])
```

Creating Model

```python
model = LinearRegression()
model.fit(X,Y)
```

Accuracies

```python
Ypred = model.predict(X)
print(r2_score(Y,Ypred))
print(mean_squared_error(Y,Ypred))
```

```python
print(df['Output'].iloc[48],np.exp(Ypred[48]))
print(df['Output'].iloc[103],np.exp(Ypred[103]))
print(df['Output'].iloc[258],np.exp(Ypred[258]))
```
